# API

API (Application Programming Interface) är ett koncept som förklarar hur applikationer kan prata med varandra. I det här kapitlet ska vi utforska hur vi kan få program att prata med olika tjänster på internet för att få tag i data.

````python
import requests

# 1. Definiera URL:en till API-tjänsten (Stockholms koordinater)
url = "https://api.open-meteo.com/v1/forecast?latitude=59.3293&longitude=18.0686&current_weather=true"

# 2. Skicka frågan till servern
respons = requests.get(url)

# 3. Omvandla svaret från JSON till en Python-dictionary
data = respons.json()

# 4. Hämta ut specifik information ur vår dictionary
aktuellt_vader = data["current_weather"]
temperatur = aktuellt_vader["temperature"]
vind = aktuellt_vader["windspeed"]

# 5. Presentera resultatet för användaren
print(f"Väderrapport för Stockholm:")
print(f"---")
print(f"Temperatur: {temperatur}°C")
print(f"Vindhastighet: {vind} m/s")
````

I ovanstående exempel frågar vi tjänsten [meteo](https://open-meteo.com/) om väderdata i Stockholm. Börja med att öppna webbsidan som laddas in i *url* variabeln och kolla hur den ser ut:

## Nerbrytning av kod

### Steg 1
Det första är att vi laddar in url:en i en variabel som vi kallar *url*. *url* är ett annat ord för addressen till hemsidan som vi ska kalla på.

### Steg 2
I steg 2 använder vi programbiblioteket *requests* för att skicka en fråga till hemsidan. Svaret lägger vi i variabeln *respons*

### Steg 3
*respons* är JSON, för att använda datan i python så gör vi om JSON till en *dictionary* (ordbok) som vi enklare kan använda i python.

### Steg 4
Nu när vi har datan i en *dictionary* så kan vi använda den för att få ut den data som vi vill ha. Här vill jag att ni kollar på programmet och jämför det med den datan som ni kan se på hemsidan för url:en. Här är det viktigt att ni ser **kopplingen mellan kod och det som syns på hemsidan**

### Steg 5
Det sista vi gör är att printa ut resultaten av väderdata för användaren
