# API

API (Application Programming Interface) är ett koncept som förklarar hur applikationer kan prata med varandra. I det här kapitlet ska vi utforska hur vi kan få program att prata med olika tjänster på internet för att få tag i data.

````python
import requests

# 1. Definiera URL:en till API-tjänsten (Stockholms koordinater)
url = "https://api.open-meteo.com/v1/forecast?latitude=59.3293&longitude=18.0686&current_weather=true"

# 2. Skicka frågan till servern
respons = requests.get(url)

# 3. Omvandla svaret från JSON till en Python-dictionary
data = respons.json()

# 4. Hämta ut specifik information ur vår dictionary
aktuellt_vader = data["current_weather"]
temperatur = aktuellt_vader["temperature"]
vind = aktuellt_vader["windspeed"]

# 5. Presentera resultatet för användaren
print(f"Väderrapport för Stockholm:")
print(f"---")
print(f"Temperatur: {temperatur}°C")
print(f"Vindhastighet: {vind} m/s")
````

## Få ut position av ISS

Nästa exempel liknar väder exemplet ganska mycket. Vi hämtar positionen för ISS (International Space Station) just nu

````python
import requests

# 1. URL till rymdstationens live-data
url = "http://api.open-notify.org/iss-now.json"

# 2. Gör anropet och omvandla till en dictionary
respons = requests.get(url)
data = respons.json()

# 3. Hämta ut koordinaterna och tidsstämpeln
position = data["iss_position"]
latitud = position["latitude"]
longitud = position["longitude"]
status = data["message"]

# 4. Skriv ut rymdstationens position
if status == "success":
    print(" ISS POSITION ")
    print(f"Latitud (Breddgrad): {latitud}")
    print(f"Longitud (Längdgrad): {longitud}")
else:
    print("Kunde inte hämta kontakt med rymdstationen.")
````
